```python
import json
import os

folder_name = 'tolokers'

# Define directories
input_dir = f'/home/ubuntu/conversation-data/dataset-01-convai/convai/data/04_emotion_labeled_data/{folder_name}'
output_base_dir = f'/home/ubuntu/conversation-data/dataset-01-convai/convai/data/05_final_data/{folder_name}'

# Get list of JSON files in the input directory
json_files = [f for f in os.listdir(input_dir) if f.endswith('.json')]

for json_file in json_files:
    # Define file paths
    input_file_path = os.path.join(input_dir, json_file)
    output_dir = os.path.join(output_base_dir, json_file.replace('.json', ''))

    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)

    # Read the input JSON file
    with open(input_file_path, 'r', encoding='utf-8') as file:
        data = json.load(file)

    # Save participant to a separate file
    participant = data.get('participant_persona', {})
    participant_file_path = os.path.join(output_dir, 'participant.json')
    with open(participant_file_path, 'w', encoding='utf-8') as file:
        json.dump(participant, file, ensure_ascii=False, indent=4)

    # Save each message to a separate file with an incrementing ID
    messages = data.get('messages', [])
    for idx, message in enumerate(messages, start=1):
        message = {'id': f"{idx:05d}", **message}
        message_file_path = os.path.join(output_dir, f'message_{idx:05d}.json')
        with open(message_file_path, 'w', encoding='utf-8') as file:
            json.dump(message, file, ensure_ascii=False, indent=4)
```

```python
from openai import OpenAI
from openai import RateLimitError
from openai import APITimeoutError
from openai import APIConnectionError
from openai import APIError
from openai import AuthenticationError

import json
import os
from typing import Dict, List
import re
import shutil
import time

MODEL = "gpt-4o-mini"
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)

dataset = "convai"
data_state = "filled_data"
folder_name = "intermediate"
file_number = "00053"

input_directory_path = f"/home/ubuntu/conversation-data/dataset-01-convai/convai/data/except/{data_state}_{folder_name}_{file_number}_input/"
output_directory_path = f"/home/ubuntu/conversation-data/dataset-01-convai/convai/data/except/{data_state}_{folder_name}_{file_number}_output/"



def label_emotion(participant_data:str, message_data: str, message_log_data: Dict) -> str:

    # step 1. generate prompt
    id = message_data.get('id', {})
    content = message_data.get('content', {})
    if not content.get('emotion_scores') and 'text' in content:
        text = content['text']
        prompt = f"""Analyze the following text and provide emotion_scores field for the following categories: Anger, Fear, Joy, Sadness, Surprise, Love, Boredom, Neutral. The output should be in JSON format with the emotion categories as keys and their respective scores as values, totaling exactly 100. 

        # Guidelines:
        1. Consider the Participant Information and Conversation Log when interpreting the emotional content of the text.
        2. Subtle emotional cues should be reflected in the corresponding emotion scores, but don't overinterpret.
        3. Assign a very high score to that emotion category when there is clear and strong evidence of a specific emotion in the text.

        # Expected output example:
        "emotion_scores": {{
            "Anger": ,
            "Fear": ,
            "Joy": ,
            "Sadness": 
            "Surprise": ,
            "Love": ,
            "Boredom": ,
            "Neutral": 
        }},
        "text": "{text}"

        # Participant Information: 
        {participant_data},

        # Conversation Log:
        {message_log_data}

        Ensure that your scoring reflects the intensity and clarity of the emotional expression in the text."""

        print(prompt)
        print("=====================================")

        # step 2. call OpenAI API
        retries = 3
        for attempt in range(retries):
            try:
                start_time = time.time()
                response = client.chat.completions.create(
                            model=MODEL,
                            messages=[
                                {"role": "system", "content": "You are an AI assistant that helps to build conversation data set."},
                                {"role": "user", "content": prompt}
                            ],
                            temperature=0.8,
                            response_format={"type": "json_object"}
                        )
                end_time = time.time()
            
                print(f"message id: {id}")
                print(f"message: {text}")
                print(f"API call took {end_time - start_time} seconds")
                print("=====================================")

                response_content = response.choices[0].message.content
                print(f"response_content: {response_content}")
                print("=====================================")

                try:
                    message_data = json.loads(response_content)  
                except json.JSONDecodeError as e:
                    print(f"JSONDecodeError: {e}")
                    continue

                if content.get('id') == id:
                # if content.get('text') == text: # regacy code before adding id
                    content['emotion_scores'] = message_data.get('emotion_scores', {})

                break

            except (APITimeoutError, APIConnectionError, APIError, AuthenticationError) as e:
                print(f"Attempt {attempt + 1} of {retries} failed with {e}. Retrying...")
                time.sleep(3) 
        else:
            print(f"Failed to process message: {text}")

    return message_data

def process_files(input_directory_path: str, output_directory_path: str):

    # load data from input directory

    # participant data
    participant_filename = "participant.json"
    participant_file_path = os.path.join(input_directory_path, participant_filename)

    with open(participant_file_path, 'r', encoding='utf-8') as file:
        participant_data = json.load(file)
    print(participant_data)
    print("---------------------------------")

    # message data
    message_files = sorted([f for f in os.listdir(input_directory_path) if f.startswith("message") and f.endswith(".json")])
    message_log_data = []

    for message_filename in message_files:
        message_file_path = os.path.join(input_directory_path, message_filename)
        with open(message_file_path, 'r', encoding='utf-8') as file:
            message_data = json.load(file)
            message_log_data.append(message_data)

            # Keep only the latest 100 messages
            if len(message_log_data) > 100:
                message_log_data.pop(0)

            print(f"message_data: {message_data}")
            print(f"message_log_data: {message_log_data}")
            print("=====================================")

            # call label emotion function
            start_time = time.time()
            message_data = label_emotion(participant_data, message_data, message_log_data)
            end_time = time.time()

            print(f"Processing {message_filename} took {end_time - start_time} seconds")
            print(f"final_data_{message_filename}:", message_data)
            print("=====================================")

            # save the updated message data to the output directory
            shutil.copy(participant_file_path, output_directory_path) # copy participant.json to the output directory
            output_message_file_path = os.path.join(output_directory_path, message_filename)

            try: 
                with open(output_message_file_path, 'w', encoding='utf-8') as file:
                    json.dump(message_data, file, ensure_ascii=False, indent=4)

                print(f"participant.json copied to {output_directory_path}")
                print(f"{message_filename} saved to {output_message_file_path}")
            
            except:
                print(f"Failed to save {message_filename} to {output_message_file_path}")     
                continue     

    return

# function call
process_files(input_directory_path, output_directory_path)
```

```python
import json
import os
from typing import List, Dict
from openai import OpenAI, RateLimitError, APITimeoutError, APIConnectionError, APIError, AuthenticationError
import time

MODEL = "gpt-4o-mini"
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)

def check_empty_emotion_scores(directory_path: str) -> List[str]:
    """
    Check for message files with empty emotion_scores in the given directory.
    
    :param directory_path: Path to the directory containing message files
    :return: List of filenames with empty emotion_scores
    """
    empty_scores_files = []
    for filename in os.listdir(directory_path):
        if filename.startswith("message") and filename.endswith(".json"):
            file_path = os.path.join(directory_path, filename)
            with open(file_path, 'r', encoding='utf-8') as file:
                message_data = json.load(file)
                if 'content' in message_data and 'emotion_scores' not in message_data['content']:
                    empty_scores_files.append(filename)
    
    empty_scores_files = sorted(empty_scores_files)
    return empty_scores_files

def reprocess_empty_emotion_scores(input_directory_path: str, output_directory_path: str, empty_files: List[str]):
    """
    Reprocess files with empty emotion_scores.
    
    :param input_directory_path: Path to the input directory
    :param output_directory_path: Path to the output directory
    :param empty_files: List of filenames to reprocess
    """
    participant_file_path = os.path.join(input_directory_path, "participant.json")
    with open(participant_file_path, 'r', encoding='utf-8') as file:
        participant_data = json.load(file)

    message_log_data = []

    for filename in empty_files:
        file_path = os.path.join(input_directory_path, filename)
        with open(file_path, 'r', encoding='utf-8') as file:
            message_data = json.load(file)
            message_log_data.append(message_data)

            if len(message_log_data) > 100:
                message_log_data.pop(0)

            updated_message_data = label_emotion(participant_data, message_data, message_log_data)

            output_file_path = os.path.join(output_directory_path, filename)
            with open(output_file_path, 'w', encoding='utf-8') as outfile:
                json.dump(updated_message_data, outfile, ensure_ascii=False, indent=4)

            print(f"Reprocessed and saved: {filename}")

def label_emotion(participant_data:str, message_data: str, message_log_data: Dict) -> Dict:

    # step 1. generate prompt
    id = message_data.get('id', {})
    role = message_data.get('role', {})
    content = message_data.get('content', {})
    if not content.get('emotion_scores') and 'text' in content:
        text = content['text']
        prompt = f"""Analyze the following text and provide emotion_scores field for the following categories: Anger, Fear, Joy, Sadness, Surprise, Love, Boredom, Neutral. The output should be in JSON format with the emotion categories as keys and their respective scores as values, totaling exactly 100. 

        # Guidelines:
        1. Consider the Participant Information and Conversation Log when interpreting the emotional content of the text.
        2. Subtle emotional cues should be reflected in the corresponding emotion scores, but don't overinterpret.
        3. Assign a very high score to that emotion category when there is clear and strong evidence of a specific emotion in the text.

        # Expected output example:
        "emotion_scores": {{
            "Anger": ,
            "Fear": ,
            "Joy": ,
            "Sadness": 
            "Surprise": ,
            "Love": ,
            "Boredom": ,
            "Neutral": 
        }},
        "text": "{text}"

        # Participant Information: 
        {participant_data},

        # Conversation Log:
        {message_log_data}

        Ensure that your scoring reflects the intensity and clarity of the emotional expression in the text."""

        print(prompt)
        print("=====================================")

        # step 2. call OpenAI API
        retries = 3
        for attempt in range(retries):
            try:
                start_time = time.time()
                response = client.chat.completions.create(
                            model=MODEL,
                            messages=[
                                {"role": "system", "content": "You are an AI assistant that helps to build conversation data set."},
                                {"role": "user", "content": prompt}
                            ],
                            temperature=0.8,
                            response_format={"type": "json_object"}
                        )
                end_time = time.time()
            
                print(f"message id: {id}")
                print(f"message: {text}")
                print(f"API call took {end_time - start_time} seconds")
                print("=====================================")

                response_content = response.choices[0].message.content
                print(f"response_content: {response_content}")
                print("=====================================")

                try:
                    response_content = json.loads(response_content)  
                except json.JSONDecodeError as e:
                    print(f"JSONDecodeError: {e}")
                    continue

                if 'content' not in message_data or not isinstance(message_data['content'], dict):
                    return message_data

                if message_data.get('content', {}).get('text') == text:
                    message_data['content']['emotion_scores'] = response_content['emotion_scores']
                    print(f"content: {content}")
                    print(f"Updated message_data: {message_data}")
                    print("=====================================")

                break

            except (APITimeoutError, APIConnectionError, APIError, AuthenticationError) as e:
                print(f"Attempt {attempt + 1} of {retries} failed with {e}. Retrying...")
                time.sleep(3) 
        else:
            print(f"Failed to process message {id}: {text}")

    return message_data

def process_directory(input_directory_path: str, output_directory_path: str):
    """
    Process a single directory, checking for empty emotion scores and reprocessing if necessary.
    
    :param input_directory_path: Path to the input directory
    :param output_directory_path: Path to the output directory
    """
    while True:
        empty_files = check_empty_emotion_scores(input_directory_path)
        if not empty_files:
            print(f"All files in {input_directory_path} have emotion scores.")
            break
        
        print(f"Found {len(empty_files)} files with empty emotion scores. Reprocessing...")
        reprocess_empty_emotion_scores(input_directory_path, output_directory_path, empty_files)

def process_all_folders(base_directory_path: str, dataset: str, data_state: str, folder_name: str):
    """
    Process all folders in the base directory.
    
    :param base_directory_path: Path to the base directory containing all subfolders
    :param dataset: Dataset name
    :param data_state: Data state
    :param folder_name: Folder name
    """
    for item in os.listdir(base_directory_path):
        if item.startswith(f"{dataset}_{data_state}_{folder_name}_"):
            input_directory_path = os.path.join(base_directory_path, item)
            output_directory_path = input_directory_path  # Same as input in this case
            
            if os.path.isdir(input_directory_path):
                print(f"Processing directory: {input_directory_path}")
                process_directory(input_directory_path, output_directory_path)

# Main execution
if __name__ == "__main__":
    
    dataset = "convai"
    data_state = "emotion_labeled_data"
    folder_name = "summer_wild_evaluation_dialogs"
    base_directory_path = f"/home/ubuntu/conversation-data/dataset-01-convai/convai/data/05_final_data/{folder_name}"

    process_all_folders(base_directory_path, dataset, data_state, folder_name)
```
